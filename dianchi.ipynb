```python
import numpy as np
import pandas as pd
import torch#（张量）：在 PyTorch 里，所有数据都必须转成 Tensor 才能喂给模型。它会用GPU，有自动求导机制
#你可以这么理解：模型在猜结果（比如猜电压是 3.2V），真实值是 3.3V。误差就是差的那 0.1V。
#模型内部有无数个旋钮（参数）。自动求导（Autograd）就像是一个“找方向仪”。它通过微积分里的“链式法则”，自动算出：“如果我调高A旋钮，误差是变大还是变小？调低B旋钮，误差会怎么变？”
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler#（归一化）
#神经网络就像个“极其挑剔的机器”。
#你的电压是 3.3V，时间是 1000小时，温度是 60℃。如果不做处理直接丢给神经网络，它会因为数字太大，导致内部的计算爆炸（专业叫“梯度爆炸”），根本学不会。
#归一化就是“把所有数据按比例压缩到 0 到 1 之间”。公式很简单：(当前值 - 最小值) / (最大值 - 最小值)。
import matplotlib.pyplot as plt

#算完之后，优化器（Optimizer）就会顺着“误差变小的方向”去拧这些旋钮。这就是模型学习的本质。

# 假设 df 已经是你读好的 FC1 数据表
# 1. 提取我们要的列（先用单变量电压试试水）
data = df['Utot (V)'].values.reshape(-1, 1)#提取数字，变为-1行（不存在，让机子自己算），1列
#加上 .values，就是把这列数据剥掉“Excel外衣”，变成纯纯的 NumPy 数组（类似于 [3.3, 3.31, 3.29...]）。神经网络不认 pandas 格式，只认纯数字数组。
#变成 (-1, 1) 的意思是：变成 1 列，行数让电脑自己算（-1 就代表自动推算）。

# 2. 归一化（训练技巧：神经网络对数字大小极度敏感，3.3V 进去容易梯度爆炸，变成 0-1 之间最好）
scaler = MinMaxScaler(feature_range=(0, 1))#scaler：归一化
#这行代码只是定义了这个“0~1”压缩工具，还没有真正去压缩数据。
data_scaled = scaler.fit_transform(data)#之所以叫 _scaled，只是为了方便程序员看懂
#scaler归一化；fit：计算数据的最大值和最小值；transform：把所有数字套进公式 (当前值 - 最小值) / (最大值 - 最小值)。

# 3. 构建滑动窗口（核心逻辑！）
def create_sequences(data, seq_length):#假设数据是：[10, 20, 30, 40, 50, 60]，seq_length = 3（用过去 3 个预测第 4 个）。
    xs = []
    ys = []
    for i in range(len(data) - seq_length):
        # x 是过去 seq_length 天的数据
        x = data[i : i + seq_length]
        # y 是紧接着的下一天的数据
        y = data[i + seq_length]
        xs.append(x)
        ys.append(y)
    return np.array(xs), np.array(ys)

SEQ_LENGTH = 24 # 用过去 24 个点预测下一个点
X_lstm, y_lstm = create_sequences(data_scaled, SEQ_LENGTH)

# 4. 划分训练集和测试集（必须按时间顺序！80% 训练）
train_size = int(len(X_lstm) * 0.8)
X_train_lstm, X_test_lstm = X_lstm[:train_size], X_lstm[train_size:]
y_train_lstm, y_test_lstm = y_lstm[:train_size], y_lstm[train_size:]

# 5. 转成 PyTorch 专属的张量 (Tensor)
# LSTM 要求的形状：(Batch_Size, Sequence_Length, Features)
X_train_tensor = torch.FloatTensor(X_train_lstm)
y_train_tensor = torch.FloatTensor(y_train_lstm)
X_test_tensor = torch.FloatTensor(X_test_lstm)
y_test_tensor = torch.FloatTensor(y_test_lstm)

print("X_train 形状:", X_train_tensor.shape) # 应该是 [52000多, 24, 1]
print("y_train 形状:", y_train_tensor.shape) # 应该是 [52000多, 1]

# 定义 LSTM 模型
class VoltageLSTM(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2, output_size=1):
        #支持一个接口（input_size=1），内存（hidden_size=64），两层显卡（num_layers=2）
        super(VoltageLSTM, self).__init__()
        # 1. LSTM 层（核心层）
        # input_size=1 (只输入电压), hidden_size=64 (大脑里记 64 个特征，越多越聪明也越容易过拟合)
        # num_layers=2 (两层叠起来，学习更深的规律)
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)

        # 2. 全连接层（把 LSTM 记住的东西，翻译成我们想要的 1 个预测数值）
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        # x 的形状是 (batch_size, seq_length, features)
        # out 包含了所有时间步的输出，_ 是隐藏状态（我们这里不需要用它）
        out, _ = self.lstm(x)

        # 我们只关心最后一个时间步（即第 24 个点）推导出来的结果
        # out[:, -1, :] 的意思就是取所有 batch 的最后一个序列的输出
        out = self.fc(out[:, -1, :])
        return out

# 实例化模型
model = VoltageLSTM()
print(model)

 # 1. 定义损失函数和优化器
criterion = nn.MSELoss() # 均方误差（之前我们讲过的那个）
# Adam 是最好用的优化器。lr=0.001 是学习率（步子迈多大）
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# 2. 开始训练
epochs = 50 # 训练 50 轮
# 训练技巧：记录 loss，画图看它是不是在下降
loss_history = []

model.train() # 开启训练模式
for epoch in range(epochs):
    # 前向传播
    outputs = model(X_train_tensor)
    loss = criterion(outputs, y_train_tensor)

    # 反向传播（这三句是 PyTorch 的固定套路）
    optimizer.zero_grad() # 清空上一轮的梯度
    loss.backward()       # 根据误差，倒推责任
    optimizer.step()      # 更新模型参数

    loss_history.append(loss.item())

    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.6f}')

# 画 Loss 曲线（如果曲线一路向下，说明模型学得很开心）
plt.plot(loss_history)
plt.title('Training Loss over Epochs')#Training Loss over Epochs:误差成绩下降曲线
plt.xlabel('Epoch')#Epochs（轮次）Loss（损失/误差）
plt.ylabel('Loss (MSE)')#横轴是学了第几遍，纵轴是误差。
plt.show()

# 1. 测试阶段
model.eval() # 开启测试模式
with torch.no_grad(): # 测试时不需要计算梯度，节省内存
    predictions = model(X_test_tensor)

# 2. 反归一化（我们之前把 3.3V 变成了 0.1，现在要变回来才能和真实值比）
predictions_real = scaler.inverse_transform(predictions.numpy())
y_test_real = scaler.inverse_transform(y_test_tensor.numpy())

# 3. 画图对比
plt.figure(figsize=(12, 5))
# 画真实的后 20% 数据
plt.plot(range(len(y_test_real)), y_test_real, label='True Voltage', color='gray', alpha=0.6)
# 画 LSTM 预测的数据
plt.plot(range(len(predictions_real)), predictions_real, label='LSTM Prediction', color='blue')
plt.xlabel('Time Step (in test set)')
plt.ylabel('Utot (V)')
plt.title('LSTM Voltage Prediction (Tail Extrapolation)')
plt.legend()
plt.grid(True)
#plt.savefig(r'C:\Users\YOGA\Desktop\神经学习v1', dpi=300, bbox_inches='tight')
plt.show()

# 算一下 RMSE
from sklearn.metrics import mean_squared_error
import math
rmse_lstm = math.sqrt(mean_squared_error(y_test_real, predictions_real))
print(f"LSTM 测试集 RMSE: {rmse_lstm:.6f} V")
```

```python
import torch
import torch.nn as nn

# ================= 物理约束模块 =================
class PhysicsInformedLoss(nn.Module):
    def __init__(self, lambda_penalty=0.1):
        super(PhysicsInformedLoss, self).__init__()
        self.mse = nn.MSELoss()
        self.lambda_penalty = lambda_penalty # 惩罚力度

    def forward(self, y_pred, y_true):
        # 1. 普通的预测误差
        mse_loss = self.mse(y_pred, y_true)

        # 2. 物理惩罚：相邻预测点之间的差值
        # y_pred 的形状是 (batch_size, 1)，我们没法直接看序列趋势
        # 为了演示，这里我们用一个小技巧：如果预测值比上一个真实值还大，就惩罚
        # （完整的单调性约束通常需要在自回归生成时按序列惩罚）

        # 简单版惩罚：预测出的电压趋势不应该剧烈向上突变
        # 因为我们已知电池在衰减，如果预测的下一个值显著高于当前值，就不符合物理规律
        # 这里我们用一个近似：惩罚相邻样本预测值的正向跳跃
        diff = y_pred[1:] - y_pred[:-1]
        # 如果 diff > 0（上升），relu 会保留它；如果下降，relu 会变成 0
        penalty = torch.mean(torch.relu(diff))

        total_loss = mse_loss + self.lambda_penalty * penalty
        return total_loss, mse_loss, penalty

# ================= 训练循环 =================
criterion = PhysicsInformedLoss(lambda_penalty=0.5) # 这个0.5可以调整，找到适合的。

epochs = 30
for epoch in range(epochs):
    model.train()
    epoch_loss = 0
    for batch_X, batch_y in train_loader:
        optimizer.zero_grad()
        outputs = model(batch_X)

        # 注意：这里用我们自定义的物理 Loss
        loss, mse, penalty = criterion(outputs, batch_y)

        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], 总Loss: {loss.item():.6f}, MSE: {mse.item():.6f}, 物理惩罚: {penalty.item():.6f}')
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, TensorDataset
import math
from sklearn.metrics import mean_squared_error

# 解决中文显示问题
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# ================= 1. 读数据与特征工程 =================
file_path = r'C:\Users\YOGA\Desktop\电池ai分析\IEEE2014DataChallengeData\IEEE 2014 Data Challenge Data\FC1_Without_Ripples\FC1.csv'
df = pd.read_csv(file_path)#（这里FC1是合并过后的表格，不需要考虑兼容性，因为这已经是转换过的了）

# 【关键升级】喂给它 4 个特征：时间、电压、出口氢温、空气湿度
# 确保这些列名在 df.columns 里存在，如果没有，你可以把温度换成 ToutWAT (°C) 等
features = ['Time (h)', 'Utot (V)', 'ToutH2 (°C)', 'HrAIRFC (%)']
data = df[features].values

split_idx = int(len(data) * 0.8)
train_data = data[:split_idx]
test_data = data[split_idx:]

# 防泄漏归一化
scaler = MinMaxScaler(feature_range=(0, 1))
train_scaled = scaler.fit_transform(train_data)
test_scaled = scaler.transform(test_data)

# ================= 2. 滑动窗口 =================
def create_sequences(data, seq_length):
    xs, ys = [], []
    for i in range(len(data) - seq_length):
        xs.append(data[i : i + seq_length])
        # 注意：y 取的是下标为 1 的列（也就是 Utot），因为我们要预测电压
        ys.append(data[i + seq_length, 1])
    return np.array(xs), np.array(ys)

SEQ_LENGTH = 50
X_train_lstm, y_train_lstm = create_sequences(train_scaled, SEQ_LENGTH)
X_test_lstm, y_test_lstm = create_sequences(test_scaled, SEQ_LENGTH)

X_train_tensor = torch.FloatTensor(X_train_lstm)
y_train_tensor = torch.FloatTensor(y_train_lstm).unsqueeze(1)
X_test_tensor = torch.FloatTensor(X_test_lstm)
y_test_tensor = torch.FloatTensor(y_test_lstm).unsqueeze(1)

# ================= 3. 定义 GRU 模型 =================
class VoltageGRU(nn.Module):
    # 注意：input_size=4 了！因为有四个特征
    def __init__(self, input_size=4, hidden_size=2, num_layers=1, output_size=1):
        super(VoltageGRU, self).__init__()
        # 只改了这一行：把 LSTM 换成了 GRU
        self.gru = nn.GRU(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        out, _ = self.gru(x)
        out = self.fc(out[:, -1, :])
        return out

model = VoltageGRU()
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
train_loader = DataLoader(TensorDataset(X_train_tensor, y_train_tensor), batch_size=512, shuffle=True)

import torch
import torch.nn as nn

# ================= 物理约束模块 =================
class PhysicsInformedLoss(nn.Module):
    # 增加惩罚力度和下降限制
    def __init__(self, lambda_up=0.5, lambda_down=0.5, max_step_drop=0.01):
        super(PhysicsInformedLoss, self).__init__()
        self.mse = nn.MSELoss()
        self.lambda_up = lambda_up
        self.lambda_down = lambda_down
        self.max_step_drop = max_step_drop # 归一化后，单步最大允许下降量（比如 0.01）

    def forward(self, y_pred, y_true, current_v, prev_v):
        mse_loss = self.mse(y_pred, y_true)

        # 惩罚 1：预测值不能比当前真实值高（不许上升）
        penalty_up = torch.mean(torch.relu(y_pred - current_v))

        # 惩罚 2：单步下降不能超过设定的阈值（不许暴跌）
        # 计算预测的下降幅度
        drop_pred = current_v - y_pred
        # 如果下降幅度超过了 max_step_drop，就会产生正数，被 relu 提取并惩罚
        penalty_down = torch.mean(torch.relu(drop_pred - self.max_step_drop))

        # 总损失
        total_loss = mse_loss + self.lambda_up * penalty_up + self.lambda_down * penalty_down
        return total_loss, mse_loss, penalty_up, penalty_down

# ================= 训练循环替换 =================
epochs = 20
criterion = PhysicsInformedLoss(lambda_up=0.5, lambda_down=0.5, max_step_drop=0.005)

for epoch in range(epochs):
    model.train()
    epoch_loss = 0
    for batch_X, batch_y in train_loader:
        # 提取当前的真实电压和上一步的真实电压
        current_v = batch_X[:, -1, 1].unsqueeze(1)
        prev_v = batch_X[:, -2, 1].unsqueeze(1)

        optimizer.zero_grad()
        outputs = model(batch_X)

        # 【修复】一次性把四个参数传进去
        loss, mse, p_up, p_down = criterion(outputs, batch_y, current_v, prev_v)

        loss.backward()
        optimizer.step() # 【修复】删掉重复的 backward 和 step

        epoch_loss += loss.item()

    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], 总Loss: {loss.item():.6f}, MSE: {mse.item():.6f}, 违规上升: {p_up.item():.6f}, 暴跌: {p_down.item():.6f}')

# ================= 5. 测试与反归一化 =================
model.eval()
with torch.no_grad():
    predictions = model(X_test_tensor)

# 多特征反归一化（核心难点，不要改错）
# 建一个假的 4 维矩阵，只把第 1 列（电压）换成预测值，其他填 0
dummy = np.zeros((len(predictions), 4))
dummy[:, 1] = predictions.numpy().flatten()
predictions_real = scaler.inverse_transform(dummy)[:, 1]

# 对真实值也做同样处理
dummy_y = np.zeros((len(y_test_tensor), 4))
dummy_y[:, 1] = y_test_tensor.numpy().flatten()
y_test_real = scaler.inverse_transform(dummy_y)[:, 1]

# ================= 6. 画图与评估 =================
plt.figure(figsize=(14, 6))
plt.plot(y_test_real, label='真实电压', color='gray', alpha=0.6)
plt.plot(predictions_real, label='GRU 预测', color='orange')
plt.title('多特征 GRU 测试集预测')
plt.legend()
plt.grid(True)
plt.show()

rmse_gru = math.sqrt(mean_squared_error(y_test_real, predictions_real))
print(f"🔥 多特征 GRU 测试集 RMSE: {rmse_gru:.6f} V")
```
